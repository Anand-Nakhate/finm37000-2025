1. What are the bid and ask, price and size, in the spread market implied IN by the outright legs?

$
S = H - U.
$

- To buy the spread: buy $H$ at ask, sell $U$ at bid  
  $
  S_{\text{implied bid}} = A_H - B_U = 96.605 - 96.960 = -0.355 \;(\approx -35.5).
  $

  Size:
  $
  \min(2190,\ 5563) = 2190.
  $

  S_{\text{implied ask}} = B_H - A_U = 96.60 - 96.965 = -0.365 \;(\approx -36.5).
  $

  Size:
  $

  $

- Implied bid in $SR3H6 - SR3U7$: $-35.5$ for 2190  
- Implied ask in $SR3H6 - SR3U7$: $-36.5$ for 52

  $
